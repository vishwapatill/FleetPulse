```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("vishwa").getOrCreate()
```

```python
df=spark.read.parquet("data\green_tripdata_2026-01.parquet")
```

```python
df.describe().show()
```

```python
df.printSchema()
```

```python
df=df.withColumnRenamed("lpep_dropoff_datetime","drop_time")
```

```python
from pyspark.sql.functions import hour
df.withColumn("Hour",hour("drop_time")).groupBy("Hour").avg("fare_amount").plot.bar(x='Hour',y="avg(fare_amount)")
```

```python
from pyspark.sql.functions import col
df.groupBy("PULocationID").count().sort(col("count").desc()).show(10)
```

```python
from pyspark.sql.functions import dayofweek
df.withColumn("week_day",dayofweek("drop_time")).groupBy("week_day").sum("fare_amount").sort("week_day").show()
```

```python
from pyspark.sql.functions import dayofweek ,monthname
df.withColumn("week_day",dayofweek("drop_time")).groupBy("week_day").sum("fare_amount").sort("week_day").plot.bar(x="week_day",y="sum(fare_amount)")
```

```python
df.columns
```

```python
ldf=spark.read.csv("data\\taxi_zone_lookup.csv",header=True)
```

```python
ldf.show()
```

```python
df=df.join(ldf,df.DOLocationID==ldf.LocationID,how="left")
df.groupBy("Borough").avg("fare_amount").show()
```

```python
from pyspark.sql.functions import unix_timestamp, col

df = df.withColumn(
    "trip_duration_seconds",
    -unix_timestamp("lpep_pickup_datetime") + unix_timestamp("lpep_dropoff_datetime")
)
df.groupBy("Borough").avg("trip_duration_seconds").show()
```

```python
df.withColumn("Hour",hour("lpep_pickup_datetime")).groupBy("Hour").count().show()
```

```python
from pyspark.sql.functions import date_format ,max ,min
from pyspark.sql import Window
df=df.withColumn("day_month",date_format("lpep_pickup_datetime",'dd/mm'))
```

```python
df=df.withColumn("Higest_for_day",max("fare_amount").over(Window.partitionBy("day_month")))
df=df.withColumn("lowest_for_day",min("fare_amount").over(Window.partitionBy("day_month")))
```

```python
df.groupBy("day_month").min("fare_amount").show()
```

```python
from pyspark.sql.functions import first ,avg
df.withColumn("ex",first("fare_amount").over(Window.partitionBy("day_month").orderBy("fare_amount"))).show()
```

```python
df=df.withColumn("drop_time",unix_timestamp("drop_time"))
```

```python
df.withColumn("7_days_Avg",avg("fare_amount").over(Window.orderBy("drop_time").rangeBetween(-7*24*60*60,0))).show()
```

```python
from pyspark.sql.functions import col
df.groupBy("PULocationID","DOLocationID").count().sort(col("count").desc()).show()
```

```python
from pyspark.sql.functions import dayofweek
df.withColumn("is_weekday",col("lpep_pickup_datetime").transform(lambda x: dayofweek(x).isin([1,7])  ))..groupBy("is_weekday").avg("fare_amount").show()
```

```python
from pyspark.sql.functions import count

df.groupBy(
    "PULocationID",
    "DOLocationID",
    "week_day"
).count().orderBy("count").show()
```

```python
from pyspark.sql.functions import mean, stddev, col, try_divide

window = Window.partitionBy(
    "PULocationID",
    "DOLocationID",
    "week_day"
)

df = df.withColumn(
    "mean_s",
    mean("total_amount").over(window)
)

df = df.withColumn(
    "stddev_s",
    stddev("total_amount").over(window)
)

df = df.withColumn(
    "z_score",
    try_divide(
        col("total_amount") - col("mean_s"),
        col("stddev_s")
    )
)

df.filter(col("z_score") > 3).show()
```

```python
from pyspark.sql.functions import col ,count ,when ,unix_timestamp
df.select([count(when(col(c).isNull(),c)).alias(c) for c in df.columns]).show()
```

```python
df.groupBy(df.columns).count().filter(col("count")>1).show()
```
